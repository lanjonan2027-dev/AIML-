```python
#task
#Get weather report of a specific region
#Find the hottest and coldest days
#Find average temperature
#find dates having >35 degree celcius

import csv

filename = "export.csv"

with open(filename, "r") as file:
  reader = csv.DictReader(file)

  highest_temp = float("-inf")
  lowest_temp = float("inf")

  for row in reader:
    temperature = float(row["temp"])

    if temperature > highest_temp:
      highest_temp = temperature
    elif temperature < lowest_temp:
      lowest_temp = temperature

print("hottest day:",highest_temp)
print("lowest day:",lowest_temp)


```
