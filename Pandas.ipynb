### Loading the Titanic Dataset
First, I will download the Titanic dataset and load it into a pandas DataFrame.

```python
import kagglehub
import pandas as pd
import os

path = kagglehub.dataset_download("yasserh/titanic-dataset")
print("Path ", path)
print(os.listdir(path))
```

```python
titanic_df = pd.read_csv(os.path.join(path, 'Titanic-Dataset.csv'))
titanic_df['Survived_Status'] = titanic_df['Survived'].map({0: 'Non-Survivor', 1: 'Survivor'})

average_age_by_survival = titanic_df.groupby('Survived_Status')['Age'].mean().reset_index()

print("Average age of survivors and non-survivors:")
display(average_age_by_survival)
```

### 2. List the passengers who are older than 60 years

```python
# Filter passengers older than 60
passengers_over_60 = titanic_df[titanic_df['Age'] > 60].reset_index(drop=True)

print("Passengers older than 60 years:")
display(passengers_over_60)
```

### 3. Find the average fare for each passenger class

```python
# Map 'Pclass' column for better readability
titanic_df['Passenger_Class'] = titanic_df['Pclass'].map({1: 'First Class', 2: 'Second Class', 3: 'Third Class'})

# Calculate average fare for each passenger class
average_fare_by_class = titanic_df.groupby('Passenger_Class')['Fare'].mean().reset_index()

print("Average fare for each passenger class:")
display(average_fare_by_class)
```

Then, you can use the `git config` command to store your credentials securely. Colab's Git Credential Manager handles this well.

```python
# This command will prompt you to authenticate via a web browser or paste a PAT.
# It uses the Git Credential Manager that's pre-installed in Colab.
!git config --global user.email "your_email@example.com" # Replace with your GitHub email
!git config --global user.name "Your GitHub Username" # Replace with your GitHub username

# To enable credential caching (optional, but convenient for repeated operations)
# The credential manager will handle storing your PAT securely.
!git config --global credential.helper store
print("Authentication setup complete. You will be prompted to authenticate when you perform an action requiring it.")
```
